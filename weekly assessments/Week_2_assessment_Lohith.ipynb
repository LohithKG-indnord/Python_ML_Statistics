1. NumPy: Create a NumPy array of 20 random integers (1–100). Without using any loops, print all

```python
import numpy as np
from numpy import random

x = random.randint(100, size=(20))
y = x.mean()
x = x[x>y]
print(x)
```

2. Pandas basics: Build a DataFrame of 6 employees with name, department, and salary. Filter
employees earning above 50,000, then use groupby to show the average salary per department

```python
import pandas as pd

employees = {
    "name": ["Rohit","jay","vinay","sachin","sunil","rohan"],
    "department": ["IT","HR","IT","HR","HR","IT"],
    "salary": [60,40,70,80,40,50]
}
af = pd.DataFrame(employees)
df = af.groupby("department")["salary"].mean()
print(df)
```
