**Binomial Distribution**

A binomial distribution gives the probability of getting a certain number of successes in a fixed number of independent trials, where each trial has only two possible outcomes (success or failure).

**Normal Distribution**

 A normal distribution is a continuous probability distribution that is bell-shaped and symmetric around the mean.

**Z-Score**

A z-score tells us how many standard deviations a value is above or below the mean.

Ex:

Mean = 50, standard deviation = 10, value = 70:
	​
    z = 70-50/10 = 2
