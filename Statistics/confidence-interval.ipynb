## Confidence Interval
A confidence interval gives a range of values that is used to estimate an unknown population value.

## Correlation
Correlation shows the relationship between two variables and tells us how they change in relation to each other.

## Positive Correlation
Positive correlation means that when one variable increases, the other variable tends to increase as well.

## Negative Correlation
Negative correlation means that when one variable increases, the other variable tends to decrease.

## No Correlation
No correlation means there is no clear relationship between the two variables.

## Causation
Causation means that a change in one variable directly causes a change in another variable.

## Correlation vs Causation
Correlation only shows that two variables are related. It does not mean that one variable causes the other.
