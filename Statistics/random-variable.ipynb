## Random Variable
A random variable is a variable whose value depends on the outcome of a random experiment.

## Discrete Random Variable
A discrete random variable has countable values. For example, the number of heads when tossing a coin.

## Continuous Random Variable
A continuous random variable can take any value within a range. For example, height, weight, or temperature.

## Probability Distribution
A probability distribution shows the possible values of a random variable and how likely each value is.

## Normal Distribution
Normal distribution is a bell-shaped distribution where most of the values are close to the mean.

## Binomial Distribution
Binomial distribution is used when an experiment has a fixed number of trials and each trial has two possible outcomes.

## Poisson Distribution
Poisson distribution is used to describe how often an event happens in a fixed amount of time or space.

## Uniform Distribution
Uniform distribution is a distribution where all possible outcomes have an equal chance of occurring.
