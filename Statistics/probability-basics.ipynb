**Independent/dependent Events:**

Probability: Measure of how likely a event is going to happen

Probability = no of possibility that meets my condition/Total number of possible events

Ex: Flipping a coin

P(Head)= 1/2=50%  
P(Tail)= 1/2=50%

**Independent Events**

Two events are independent if the occurrence of one event does not affect the probability of the other.

Formula:

P(AnB)=P(A)x P(B)

Ex: Tossing a coin twice. The first toss does not affect the second toss.

**Dependent Events**

Two events are dependent if the occurrence of one event affects the probability of the other.

Formula:

P(AnB)=P(A)x P(B|A)

Ex: Drawing two cards from a deck without replacement. The first card changes the probability of getting the second card.

**Universal set, subset , superset etc**

1. Union of Sets — AuB

    The union contains all elements that are in A or B or both.

        Example:
        
            A = {1,2,3}, B = {3,4,5}
            AuB={1,2,3,4,5}  

   

2. Intersection of Sets — AnB

    The intersection contains only the common elements of A and B.

        Example:

            AnB={3}  

   

3. Difference - -

    A-B means the elements that are in A but NOT in B.

       Example:

            A = {1,2,3,4}, B = {3,4,5}
            A−B={1,2}  

   

4. Universal Set — U

    The universal set contains all elements under consideration in a particular problem.

        Example:

            U={1,2,3,4,5,6,7}  

   


5. Complement of a Set

    The complement of A contains everything in the universal set that is NOT in A.

   
   
6. Subset 

    A is a subset of B if every element of A is also in B.


7. Strict Subset 

    A is a strict subset of B if every element of A is in B, but A is not equal to B.


8. Superset 

    B is a superset of A if B contains every element of A.
