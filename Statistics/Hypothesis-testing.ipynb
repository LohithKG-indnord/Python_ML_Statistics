## Hypothesis Testing
Hypothesis testing is a method used to check whether there is enough evidence to support a claim about a population.

## Null Hypothesis (H0)
The null hypothesis is the assumption that there is no significant difference or effect.

## Alternative Hypothesis (H1)
The alternative hypothesis is the assumption that there is a significant difference or effect.

## P-value
P-value tells us how likely the observed result is if the null hypothesis is true.

## Significance Level
Significance level is the limit used to decide whether a result is statistically significant.

## Z-statistic
Z-statistic tells us how far a value is from the mean in terms of standard deviations. It is commonly used when the population standard deviation is known.

## T-statistic
T-statistic is used when the population standard deviation is unknown and is commonly used with smaller samples.

## Z-statistic vs T-statistic
Z-statistic is mainly used when the population standard deviation is known, while T-statistic is used when it is unknown.
