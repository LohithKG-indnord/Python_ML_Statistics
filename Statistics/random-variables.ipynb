**Random Variables**

A random variable is a variable that assigns a numerical value to the outcome of a random experiment

Ex:
If you toss a coin twice,

let  X = number of heads

Possible values of X:0,1,2

**Types of Random Variables**

Discrete random variable: Takes countable values.  
Continuous random variable: Can take any value within a range.

**Probability Density Function:**

A probability density function is a function that describes how the probability of a continuous random variable is distributed over different values.

 **Law of Large Numbers**

The Law of Large Numbers states that as the number of trials or observations increases, the sample average gets closer to the true population mean (expected value).

**Binomial Distribution**

 A binomial distribution gives the probability of getting a certain number of successes in a fixed number of independent trials, where each trial has only two possible outcomes

**Normal Distribution**

A normal distribution is a continuous probability distribution that is bell-shaped and symmetric around the mean.
