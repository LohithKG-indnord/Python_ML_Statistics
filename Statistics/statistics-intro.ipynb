# **Descriptive Statistics**

**Average: Central Tendency**

1.Mean: sum of numbers that meet our condition/total number of elements

2.Median: The middle value when numbers are arranged in order

3.Mode: The value that occurs the most

Ex:
 num = 4, 3, 1, 6, 1, 7
 
 mean = 4+3+1+6+1+7/6 = 3.6  
 median = 1 1 3 4 6 7 = 3+4/2 = 3.5  
 mode = 1

**Range and midrange:**

Range:The difference between the highest and lowest values in a data set
    Highest value - Lowest value

Midrange:The average of the highest and lowest values.
    Mid-range = (Highest value + Lowest value) ÷ 2

Ex: list = 65,81,73,85,94,79,67,83,82

Range: 94-65 = 29  
Mid range: (94+65)/2 = 79.5

**Variance and Standard Deviation:**

Variance: The average of the squared differences between each value and the mean. It shows how spread out the data is.

Standard deviation: The square root of the variance. It shows how far the values typically are from the mean.

A box-and-whisker plot is a graph used to show the distribution and spread of data.
