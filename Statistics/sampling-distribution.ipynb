## Population
Population is the complete group of data that we want to study.

## Sample
A sample is a smaller part of a population that is selected for analysis.

## Sampling
Sampling is the process of selecting a sample from a population.

## Sampling Distribution
Sampling distribution shows how a statistic changes when we take different samples from the same population.

## Sample Mean
Sample mean is the average value calculated from a sample.

## Central Limit Theorem
Central Limit Theorem states that the distribution of sample means becomes approximately normal when sufficiently large samples are taken from a population.

## Importance of CLT
CLT helps us make conclusions about a population using samples and is useful in statistical testing and estimation.
