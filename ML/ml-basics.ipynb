# **What is machine Learning?**

    * Machine learning is the science of teaching the machines to learn by themselves

    * "Machines can do high frequency of repetitive tasks with high accuracy without getting bored or tired"

    * So machine needs a way to think to these tasks this is where machine learning models help.

# **Types of Machine Learning:**

    1.Supervised Machine Learning:
        When you have a past data outcomes(labels) and want to predict the future outcomes of a data

        There are 2 kind of problems:
            Classification: When you want to classify a given data into different classes
            Regression: Where you want to predict a continuous numerical values

    2.Unsupervised Machine Learning:
        Where you will be discovering pattern from a given data

    3.Reinforcement Learning:
        Here a agent learns to make decisions by interacting with the environment by using reward and penalty system based on its action such that it will try to maximize the rewards collected.
    

# **Steps involved in building a Model:**

    5.Modeling
